```python
LAST_NAME = "ARAGON"
STUDENT_ID = "TUPM-26-1834"
FAVORITE_ARTIST = "YOASOBI"

SEED_NUM = int(STUDENT_ID[-1])

print(f"Source Prefix: {LAST_NAME}, Seed: {SEED_NUM}")




```

```python
import random

categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]
sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]

char_multipliers = [ord(char) % 10 + 1 for char in FAVORITE_ARTIST if char.isalnum()]
random.seed(SEED_NUM)

raw_records = []

for idx, source in enumerate(sources):
    for c_idx, cat in enumerate(categories):
        # Blend random generation with favorite artist multipliers
        mult = char_multipliers[(idx + c_idx) % len(char_multipliers)]
        val = round((random.uniform(10.0, 100.0) * mult), 2)
        raw_records.append((source, cat, val))

repeated_source, repeated_cat = sources[0], categories[0]
raw_records.append((repeated_source, repeated_cat, round(random.uniform(10.0, 100.0) * char_multipliers[0], 2)))

for extra_cat in ["VOLTAGE", "SPEED"]:
    for _ in range(2):
        src = random.choice(sources)
        val = round(random.uniform(15.0, 85.0) * char_multipliers[-1], 2)
        raw_records.append((src, extra_cat, val))

print(f"Successfully generated {len(raw_records)} individualized raw records.")
```

```python
from collections import defaultdict

organized_data = defaultdict(lambda: defaultdict(list))

for source, cat, val in raw_records:
    organized_data[source][cat].append(val)

print("Data structural layout verification complete.")


```

```python
record_tracker = defaultdict(int)
distinct_pairs = set()
duplicate_pairs = set()

for source, cat, _ in raw_records:
    pair = (source, cat)
    record_tracker[pair] += 1
    distinct_pairs.add(pair)
    if record_tracker[pair] > 1:
        duplicate_pairs.add(pair)

print(f"Distinct Source-Category Tracks: {len(distinct_pairs)}")
print(f"Identified Repeated Tracks: {list(duplicate_pairs)}")
```

```python
consolidated_summary = []

for source in sorted(organized_data.keys()):
    for cat in sorted(organized_data[source].keys()):
        values = organized_data[source][cat]
        metrics = {
            "Source": source,
            "Category": cat,
            "Readings_Count": len(values),
            "All_Values": values,
            "Average_Value": round(sum(values) / len(values), 2),
            "Has_Duplicates": (source, cat) in duplicate_pairs
        }
        consolidated_summary.append(metrics)

print("Analytical synthesis and sorting pass completed.")
```

```python
print("="*80)
print(f"          LABORATORY ANALYSIS REPORT (RUN PROFILE SEED: {SEED_NUM})")
print("="*80)
print(f"\n--- ORIGINAL DATA LOG ({len(raw_records)} Records) ---")
print(f"{'Source':<18} | {'Category':<15} | {'Value':<10}")
print("-" * 50)
for src, cat, val in raw_records:
    print(f"{src:<18} | {cat:<15} | {val:<10}")

print(f"\n--- CONSOLIDATED ANALYSIS SUMMARY ---")
print(f"{'Source':<18} | {'Category':<15} | {'Count':<6} | {'Avg Value':<10} | {'Status':<12}")
print("-" * 70)
for record in consolidated_summary:
    status = "REPEATED RUN" if record["Has_Duplicates"] else "Distinct"
    print(f"{record['Source']:<18} | {record['Category']:<15} | {record['Readings_Count']:<6} | {record['Average_Value']:<10} | {status:<12}")
print("="*80)
```
