```python

LAST_NAME = "ARAGON"            
STUDENT_ID = "TUPM-26-1834"             
FAVORITE_ARTIST = "YOASOBI" 

SEED_NUM = int(STUDENT_ID[-1])


```

```python
import random

products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {
    "NOTEBOOK": 2.50,
    "PEN": 1.20,
    "USB": 15.00,
    "HEADPHONES": 25.00,
    "CALCULATOR": 12.80
}

customers = []
for i in range(1, 6):
    customer_id = LAST_NAME + "_C" + str(i)
    customers.append(customer_id)

multipliers = []
for char in FAVORITE_ARTIST:
    if char.isalnum():
        multipliers.append(ord(char) % 5 + 1) 

random.seed(SEED_NUM)
generated_transactions = []

for c_idx, customer in enumerate(customers):
    for p_idx, product in enumerate(products):
        m_offset = (c_idx + p_idx) % len(multipliers)
        mult = multipliers[m_offset]
        
        qty = random.randint(1, 4) * mult
        
        generated_transactions.append([customer, product, qty])

repeated_customer = customers[0]
repeated_product = products[0]
generated_transactions.append([repeated_customer, repeated_product, 3])

print(f"--- TRANSACTIONS LIST ({len(generated_transactions)} Records) ---")
for tx in generated_transactions:
    print(tx)
```

```python
organized_transactions = {}

for customer in customers:
    organized_transactions[customer] = {}
    for product in products:
        organized_transactions[customer][product] = []

for customer, product, qty in generated_transactions:
    organized_transactions[customer][product].append(qty)

for customer in organized_transactions:
    print(f"Customer Account: {customer}")
    for product in organized_transactions[customer]:
        quantities_list = organized_transactions[customer][product]
        if len(quantities_list) > 0:
            print(f"   └── Bought: {product:<12} => Quantities ordered: {quantities_list}")
```

```python
distinct_profiles = []
repeated_profiles = []

for customer in organized_transactions:
    for product in organized_transactions[customer]:
        history = organized_transactions[customer][product]
        
        if len(history) > 0:
            distinct_profiles.append([customer, product])
            
        if len(history) > 1:
            repeated_profiles.append([customer, product])

print(f"Total Unique (Distinct) Customer-Product Pathways: {len(distinct_profiles)}")
print(f"Total Re-purchasing Anomalies Detected: {len(repeated_profiles)}")
for track in repeated_profiles:
    print(f"  Repeated Purchase: Customer {track[0]} bought {track[1]} across multiple receipts.")
```

```python

processed_receipts = []
global_store_revenue = 0.0

for customer in sorted(organized_transactions.keys()):
    for product in sorted(organized_transactions[customer].keys()):
        all_quantities = organized_transactions[customer][product]
        
        total_qty = sum(all_quantities)
        
        if total_qty > 0:
            unit_price = prices[product]
            monetary_value = round(total_qty * unit_price, 2)
        
            global_store_revenue += monetary_value
            
            is_duplicate = [customer, product] in repeated_profiles
            
            receipt_summary = {
                "Customer": customer,
                "Product": product,
                "Total_Qty": total_qty,
                "Unit_Price": unit_price,
                "Total_Cost": monetary_value,
                "Is_Repeat_Buyer": is_duplicate
            }
            processed_receipts.append(receipt_summary)

print(f"Cumulative Gross Dataset Value: ${round(global_store_revenue, 2)}")
print("-" * 50)
for record in processed_receipts:
    print(record)
```
